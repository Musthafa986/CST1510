```python
class student:
    def __init__(self, name, DOB, program, mark):
        self.name = name
        self.DOB = DOB
        self.program = program
        self.mark = mark

    def findAge(self):
        print(self.DOB)

    def grade(self):
        if self.mark >= 40:
            print("Grade is Pass")
        else:
            print("Grade is Fail")
std1 = student("Ali", "5 Sep 2007", "AI&DS", 90)
std2 = student("Badar", "12 Feb 2008", "IT", 34)

std1.findAge()
std1.grade()

std2.findAge()
std2.grade()
```
