```python
import pandas as pd

s = pd.read_csv(r"C:\Users\mmt31\OneDrive\Desktop\MDX\CST1510\Week 6\Week 06 - Pandas\Week 06 - Pandas\03 - Grouping, Sorting and Aggregating\examples\servers.csv")
s["percent"] = (s["used"] / s["total"]) * 100

def classify(percent):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= 90:
        return "WARNING"
    else:
        return "OK"

s["status"] = s["percent"].apply(classify)
print(s)
```
